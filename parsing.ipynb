# Парсинг с помощью BeautifulSoup

Beautiful Soup - это библиотека для Python, которая позволяет парсить (анализировать) HTML и XML документы. Она предоставляет удобный способ искать, навигировать(ориентирорваться, перемещаться), и модифицировать дерево DOM (Document Object Model), представляющее HTML/XML документ.

```python
from bs4 import BeautifulSoup
```

```python
url = 'http://quotes.toscrape.com/'
```

```python
r = requests.get(url, verify=False, timeout=10)
#Этот код на Python использует библиотеку requests для выполнения HTTP-запроса типа GET к указанному url. Разберем его подробнее:

#requests.get(url, ...) — выполняет HTTP-запрос GET по адресу, указанному в переменной url.
#verify=False — отключает проверку SSL-сертификата(SSL-сертификат - это цифровой сертификат, который подтверждает подлинность веб-сайта и обеспечивает безопасное соединение между браузером пользователя и сервером). Это небезопасно и может привести к уязвимостям (например, MITM-атакам).
#timeout=10 — устанавливает максимальное время ожидания ответа в 10 секунд. Если сервер не ответит за это время, вызов вызовет исключение requests.exceptions.Timeout.
r
```

```python
soup = BeautifulSoup(r.content, 'html.parser')#Этот код использует библиотеку BeautifulSoup для парсинга HTML-ответа, полученного при выполнении HTTP-запроса. Разберем его подробнее:

#r.content — это тело ответа от сервера, которое содержится в объекте r (скорее всего, это результат запроса через requests.get()). В отличие от r.text, которое возвращает ответ в виде строки, r.content предоставляет байтовые данные, что важно, если контент может быть в другом формате (например, изображение или файл).
#BeautifulSoup(r.content, 'html.parser') — создает объект BeautifulSoup для парсинга HTML-кода, переданного в виде байтов. 'html.parser' указывает, что для парсинга будет использован стандартный парсер HTML, встроенный в Python.
#Что происходит в итоге:
#Этот код создает объект soup, который представляет собой структурированное дерево HTML-страницы, и с помощью него можно легко извлекать элементы, искать теги, атрибуты и т. д.
```

```python
print(soup.prettify())#означает:"Выведи HTML-код красиво и удобно для чтения человеком."
```

```python
divs = soup.find_all('div', attrs={'class':'quote'})#Эта строка ищет все теги <div> с классом quote на HTML-странице и сохраняет их в список divs
```

```python
df = pd.DataFrame({
    'text':[],
    'author':[],
    'author_url':[],
    'tags':[],
})
```

```python
def get_quotes(soup, data=df):#data=df означает, что если при вызове функции get_quotes не передать значение для data, то по умолчанию оно будет равно df
    
    divs = soup.find_all('div', attrs={'class':'quote'})
    
    for div in divs:
        quote = {}
        quote['text'] = div.find('span', class_='text').text
        quote['author'] = div.find('small', class_='author').text
        quote['author_url'] = div.find_all('span')[1].find('a').get('href')#Этот код находит ссылку на страницу автора цитаты и сохраняет её в словарь.
        tags = []
        for a_tag in div.find('div', class_='tags').find_all('a', class_='tag'):#Внутри блока с цитатой ищем все слова с классом 'tag' и сохраняем их в список tags
            tags.append(a_tag.text)
        quote['tags'] = ', '.join(tags)
        df.loc[len(df)] = quote #Добавляет словарь quote как новую строку в таблицу df
```

```python
url = 'https://quotes.toscrape.com/page/'

for i in range(1, 21):
    print(f'Page N:{i}')
    r = requests.get(url + str(i), timeout=10)
    soup = BeautifulSoup(r.content, 'html.parser')
    
    get_quotes(soup, df)
    
    if 'No quotes found!' in soup.text:
        break
```
